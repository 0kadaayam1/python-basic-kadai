```python
class Human:
    def __init__(self , name , age):
        self.name = name
        self.age = age

    def check_adult(self):
        print(self.name)
        print(self.age)

        if self.age < 20:
          print("未成年です")
        else:
          print("成人です")

human1 = Human("侍太郎" , 19)
human2 = Human("侍花子", 18)
human3 = Human("侍次郎", 30)

human_list = [human1 , human2 , human3]

for human in human_list:
    human.check_adult()
```
