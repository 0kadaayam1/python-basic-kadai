```python
class Human:
    def __init__(self , name , age):
        self.name = name
        self.age = age

    def check_adult(self):
        print(self.name)
        print(self.age)

        if self.age < 20:
          print("未成年です")
        else:
          print("成人です")

human = Human("侍太郎" , 19)

human.check_adult()
```
